```python
from pyspark.sql import functions as F

# Read raw trips and raw zones from Unity Catalog
trips_df = spark.table("hyf.nyc_yellow.raw_trips")
zones_df = spark.table("hyf.nyc_yellow.raw_zones")

# Q1: Which pickup borough has the most trips?
borough_trips = (
    trips_df.join(zones_df, trips_df.pickup_location_id == zones_df.location_id, "inner")
    .groupBy("borough")
    .count()
    .orderBy(F.col("count").desc())
)
borough_trips.show(10)

# Q2: Average total_amount per payment_type
payment_avg = (
    trips_df.groupBy("payment_type")
    .agg(F.avg("total_amount").alias("avg_total_amount"))
    .orderBy("payment_type")
)
payment_avg.show()
```

### When to choose PySpark vs dbt SQL
Use dbt SQL for standard analytical transformations, data modeling, and marts where SQL is concise, declarative, and easily tested. Reach for PySpark when handling unformatted raw files, complex procedural logic, machine learning pipelines, or custom algorithms that cannot be expressed cleanly in standard SQL.
